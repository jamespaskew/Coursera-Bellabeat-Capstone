## What effect does activity have on sleep

Start by exploring the data to prepare for cleaning.

```python
import pandas as pd

# Import dataset
df = pd.read_csv("sleepDay_merged.csv")

# Explore dataset
print(df.head())
print(df.info())
print(df.shape)
print(df.isnull().sum())
```

Here we see that there are no missing valure.
However the sleep day column is stored as a str and should be a datetime (for ease of future use).
Next we will convert this column to datetime.

```python
df['SleepDay'] = pd.to_datetime(
    df['SleepDay'],
    errors='coerce'
)

print(df['SleepDay'].dtype)
print("Invalid dates:", df['SleepDay'].isna().sum())
```

Next count the number of entries per ID to determine any discrepancy in participants.

```python
# Count entries per ID
id_counts = df['Id'].value_counts()

print(id_counts)
```

Here we see that one participant has 32 entries in 31 days.  This means there may be duplicate entries.
Also we see that there is a large discrepancy between the number of day they tracked their sleep
Next print all duplicates that will need to be removed.

```python
# Find all duplicate entries
duplicates = df[df.duplicated(keep=False)]

# Display all rows and columns
with pd.option_context('display.max_columns', None,
                       'display.max_rows', None):
    display(duplicates)
```

Here we see that there are 3 duplicated rows.
Next we remove these entries.

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Reset index
df = df.reset_index(drop=True)

# Confirm duplicates are removed
print("Remaining duplicates:", df.duplicated().sum())
print("Total rows:", len(df))
```

Save this in a new CVS file to preserve the original

```python
df.to_csv("sleepDay_merged_cleaned.csv", index=False)
```

## Now to explore and clean the activity file

```python
# Import dataset
df = pd.read_csv("dailyIntensities_merged.csv")

# Explore dataset
print(df.head())
print(df.info())
print(df.shape)
print(df.isnull().sum())
```

Again we see here that ActivityDay is stored as a str and not datetime
For the ease of future analysis we will convert this to datetime.

```python
df['ActivityDay'] = pd.to_datetime(
    df['ActivityDay'],
    errors='coerce'
)

print(df['ActivityDay'].dtype)
print("Invalid dates:", df['ActivityDay'].isna().sum())
```

Next we will count the number of entries per participant.

```python
# Count entries per ID
id_counts = df['Id'].value_counts()

print(id_counts)
```

Again we see that there is a descrepancy between the participants.
We do not see direct evedence of duplicants but we will check.

```python
# Find all duplicate entries
duplicates = df[df.duplicated(keep=False)]

# Display all rows and columns
with pd.option_context('display.max_columns', None,
                       'display.max_rows', None):
    display(duplicates)
```

There are no duplicated rows.
Next we check for any negative values.

```python
# Check for negative values
columns = [
    'SedentaryMinutes',
    'LightlyActiveMinutes',
    'FairlyActiveMinutes',
    'VeryActiveMinutes',
    'SedentaryActiveDistance',
    'LightActiveDistance',
    'ModeratelyActiveDistance',
    'VeryActiveDistance'
]

print((df[columns] < 0).sum())
```

There are no negative values.
Next we make sure there is not time recorded more than 24 hrs in a day.

```python
# Check activity minutes exceeding 24 hours
minute_columns = [
    'VeryActiveMinutes',
    'FairlyActiveMinutes',
    'LightlyActiveMinutes',
    'SedentaryMinutes'
]

df['TotalMinutes'] = df[minute_columns].sum(axis=1)

display(df[df['TotalMinutes'] > 1440])
```

There are no rows that sum to more than 24 hrs in a day.
